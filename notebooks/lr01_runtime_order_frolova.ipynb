**Цель:** Восстановить порядок ячеек, проверить notebook после restart.

**Мини-эксперимент:** Создадим код, в котором ячейка с print будет создана до создания переменной, что будет являться неправильным порядком, после чего поменяем порядок на правильный.

# Ошибочный порядок:

Сначала выполним ячейку с print до создания самой переменной, зафиксируем ошибку.

```python
print(project_title)
```

```python
project_title = "Mini Corpus"
```

Переменная не была выведена, код выдал ошибку - ячейки были выполнены в неправильном порядке. Код не может быть выполнен без введённой в начале переменной, действия нелогичны. При выполнении подобного кода происходит вывод несуществующей переменной.

# Правильный порядок:

Поменяем местами ячейки кода и проверим его на наличие ошибки.

```python
project_title = "Mini Corpus"
```

```python
print(project_title)
```

Команда print(project_title) не выдала ошибку в коде, ячейки были введены в правильном порядке, текст был успешно выведен на экран.

# Вывод о состоянии:

Для корректной работы ячейки должны выполняться в определённом порядке, иначе код выдаст ошибку, что было показано в "ошибочном порядке". Переменная для начала должна быть введена в код, чтобы после её вывести.
